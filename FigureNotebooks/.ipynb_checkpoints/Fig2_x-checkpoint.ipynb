```python
fig_dir="/Users/xushuo/work/Papers/Histrich/Figure/"
data_dir='/Users/xushuo/work/Project/Simulation/'
```

```python
with open(data_dir+"galaxies_tng300_072_ell_aperture_0_3.txt",'rb') as f: 
    tab=pickle.load(f)
```

```python
mask=(tab['catsh_primary']==True)&(tab['proj']=='xy')
```

```python
tab1=tab[mask]
```

```python
mask1=(tab1['mass_stellar_ins']>1e10)&(tab1['mass_stellar_exs']>10**10.5)&(tab1['c_200c']<20)&(tab1['acc_rate']<20)
```

```python
tab2=tab1
```

```python
def topNselect(data,name,number):
    data_meta=data.group_by(name)
    data0=data_meta[-number::]
    return data0
```

```python
def topNoutselect(data,name1,name2,number):
    data_meta=data.group_by(np.log10(data[name1]-data[name2]))
    data0=data_meta[-number::]
    return data0
```

```python
fig=plt.figure(figsize=(48,36))
gs = fig.add_gridspec(3,4, hspace=0, wspace=0.1)
axlist=gs.subplots(sharex='col')
label_list=[r'$\log[M_{\star,30}/M_\odot]$',r'$\log[M_{\star,100}/M_\odot]$',r'$\log[M_{\star,[50,100]}/M_\odot]$']
name_list=['aper_30_gal','aper_100_gal','out_50_100_gal']
top_numb=500
rich='richness_cut_8.5'
xlabel=r'$\log\lambda_{\log M_\star>8.5}$'
for ii in range(3):
    name=name_list[ii]
    label=label_list[ii]
    (ax1,ax2,ax3,ax4)=axlist[ii]
    if ii==2:
        ydata=np.log10(tab2[mask1]['aper_100_gal']-tab1[mask1]['aper_50_gal'])
        topNout=topNoutselect(tab2[mask1],'aper_100_gal','aper_50_gal',top_numb)
        mask_l=(ydata>np.min(np.log10(topNout['aper_100_gal']-topNout['aper_50_gal'])))
    else:
        ydata=np.log10(tab2[mask1][name])
        mask_l=(ydata>np.min(np.log10(topNselect(tab2[mask1],name,top_numb)[name]))) 
    xdata=np.log10(tab2[mask1][rich])
    mask_h=(xdata>np.min(np.log10(topNselect(tab2[mask1],rich,top_numb)[rich]))) 
    mask_up=mask_l&(~mask_h)
    mask_down=mask_h&(~mask_l)
    ax1.scatter(xdata[mask_down],ydata[mask_down],alpha=0.5,color='steelblue')
    ax1.scatter(xdata[mask_up],ydata[mask_up],alpha=0.5,color='orangered')
    
    ax1.set_ylabel(label)

    ax2.hist(np.log10(tab2[mask1][mask_down]['mass_halo']),bins=20,histtype='step',density=True,color='steelblue')
    ax2.axvline(np.log10(tab2[mask1][mask_down]['mass_halo']).mean(), color='steelblue', linestyle='--', linewidth=1.4,label=r'\rm Mean')
    ax2.axvline(np.median(np.log10(tab2[mask1][mask_down]['mass_halo'])), color='steelblue', linestyle='-.', linewidth=1.4,label=r'\rm Median')
    ax2.hist(np.log10(tab2[mask1][mask_up]['mass_halo']),bins=20,histtype='step',density=True,color='orangered')
    ax2.axvline(np.log10(tab2[mask1][mask_up]['mass_halo']).mean(), color='orangered', linestyle='--', linewidth=1.4)
    ax2.axvline(np.median(np.log10(tab2[mask1][mask_up]['mass_halo'])), color='orangered', linestyle='-.', linewidth=1.4)
    ax3.hist(tab2[mask1][mask_down]['c_200c'],bins=20,histtype='step',density=True,color='steelblue')
    ax3.axvline(tab2[mask1][mask_down]['c_200c'].mean(), color='steelblue', linestyle='--', linewidth=1.4,label=r'\rm Mean')
    ax3.axvline(np.median(tab2[mask1][mask_down]['c_200c']), color='steelblue', linestyle='-.', linewidth=1.4,label=r'\rm Median')
    ax3.hist(tab2[mask1][mask_up]['c_200c'],bins=20,histtype='step',density=True,color='orangered')
    ax3.axvline(tab2[mask1][mask_up]['c_200c'].mean(), color='orangered', linestyle='--', linewidth=1.4)
    ax3.axvline(np.median(tab2[mask1][mask_up]['c_200c']), color='orangered', linestyle='-.', linewidth=1.4)
    ax4.hist(tab2[mask1][mask_down]['acc_rate'],bins=20,histtype='step',density=True,color='steelblue')
    ax4.axvline(tab2[mask1][mask_down]['acc_rate'].mean(), color='steelblue', linestyle='--', linewidth=1.4)
    ax4.axvline(np.median(tab2[mask1][mask_down]['acc_rate']), color='steelblue', linestyle='-.', linewidth=1.4)
    ax4.hist(tab2[mask1][mask_up]['acc_rate'],bins=20,histtype='step',density=True,color='orangered')
    ax4.axvline(tab2[mask1][mask_up]['acc_rate'].mean(), color='orangered', linestyle='--', linewidth=1.4)
    ax4.axvline(np.median(tab2[mask1][mask_up]['acc_rate']), color='orangered', linestyle='-.', linewidth=1.4)
    if ii==2:
        ax1.set_xlabel(xlabel)
        ax2.set_xlabel(r'$\log[M_{\rm halo}/M_\odot]$')
        ax3.set_xlabel(r'$c_{200c}$')
        ax4.set_xlabel(r'$\ln M/\ln a$')
    if ii==0:
        ax1.text(0.05,0.9,r'\rm Top %(a)d difference set'%{'a':top_numb},transform=ax1.transAxes)
        ax2.legend()

plt.savefig(fig_dir+"Fig1.png",dpi=150)
```
